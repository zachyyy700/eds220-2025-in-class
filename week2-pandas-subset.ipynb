# Pandas Subsetting

```python
import pandas as pd

# Read in file, argument is the file path
df = pd.read_csv('data/wetlands_seasonal_bird_diversity.csv')
```

```python
# View first five rows
df.head()
```

```python
# View datatypes of columns
df.dtypes
```

## By Columns

### A single column

```python
# Select mul_spring column from df
mul_spring = df['MUL_spring']

# Print object typ
print(type(mul_spring))
```

### Multiple columns

```python
tje_wf = df[['TJE_winter', 'TJE_fall']]

print(tje_wf.head())
print(type(tje_wf))
```

### With slice

`.loc[]`: inclusive on both sides of []

```python
csm_mul = df.loc[:, 'CSM_winter':'MUL_fall']
# Print original df head
print(df.columns)
# Print sliced head
print(csm_mul.columns)
```

## By Rows

### By a condition

```python
post_2020 = df[df['year'] > 2020]
post_2020.head()
```

```python
subset0 = df[df['SDW_spring'] >= 75]
subset0
```

`.between()`

- **don't use `.loc()`** for "slicing" rows

```python
subset = df[df['year'].between(2016, 2020)]
subset.head()
```

### By multiple conditions

```python
df[(df['CSM_winter'] > 50) | (df['CSM_fall'] > 50)]
```

```python
df[(df['CSM_spring'] > 30) & (df['CSM_fall'] > 45)]
```

### By position

`.iloc()`

```python
df.iloc[6:8]
```

## By Columns & Rows

### By conditions

`df.loc[row-selection , column-selection]`

```python
df.loc[df['year'] > 2020, ['MUL_winter','MUL_spring', 'MUL_fall']]
```

### By positions

`df.iloc[row-indices, column-indices`]

- *row-indices*: **slice** of integer indices using `:`

- *column-indices*: **list** of integer indices using `,`

```python
df.iloc[6:8, [6,7]]
```
