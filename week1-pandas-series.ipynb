## Series Check-in

The integer number -999 is often used to represent missing values. Create a `pandas.Series` named `s` with four integer values, two of which are -999. The index of this series should be the the letters A through D.

```python
s = pd.Series([-999, -999, 67, 41], index=['A', 'B', 'C', 'D'])
print(s)
```

Use the `mask()` method to update the series `s` so that the -999 values are replaced by NA values. HINT: check the first example in the method’s documentation.

```python
s.mask(s == -999, inplace=True)
print(s)
```

## Dataframes Check-in

Update the column names to C1 and C2 by updating the `df.columns` attribute.

```python
# Initialize dictionary with columns' data 
d = {'col_name_1' : pd.Series(np.arange(3)),
     'col_name_2' : pd.Series([3.1, 3.2, 3.3]),
     }

# Create data frame
df = pd.DataFrame(d)
df
```

```python
df.columns = ['C1', 'C2']
df
```
