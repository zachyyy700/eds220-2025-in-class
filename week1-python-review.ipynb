# Python Review

## Libraries and packages

**library:** a collection of code to perform tasks

```python
import numpy as np
```

## Variables

**variable:** a name we assign to an object

```python
# Assign array to variable a
a = np.array([[1,1,2], [3,5,8]])
a
```

## Objects

**objects:** bundle of *properties* and *actions* 

Ex.

Object: data frame

Properties: # of rows, # of cols, date created

Actions: Select row, add column

## Types

Every object has a **type**. We can call the **type** or **class**

```python
print(a)
type(a)
```

```python
print(a[0,0])
type(a[0,0])
```

```python
print(a[1,1])
```

## Functions

Functions take in arguement. Optional & required arguments create an output.

```python
# Function help
?print
```

## Attributes & Methods

Objects have attributes and methods

- **attribute**: a property of the object, some piece of information about it

- **method**: a procedure associated with the object, an action stemming from the object

```python
print(a)

# Example attribute 
a.shape
# Example method
a.max()
```
